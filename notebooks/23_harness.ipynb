### Part 0: the chat (`services/chat/`)
`session.py` is the shared core: `ChatSession` logs in with the session token (`channel.login`), is bound to one mandate for its life, assembles context from that mandate's files only, and refuses to proceed if any other mandate's name, term or canary reaches the context. `cli.py` and `web.py` are thin front ends over it.

```
PYTHONPATH=src python -m services.chat.cli --mandate mandate_a      # token from LANTERN_SESSION_TOKEN or a hidden prompt
PYTHONPATH=src python -m services.chat.web --port 8023              # http://127.0.0.1:8023/
```

````python
sys.path.insert(0, str(ROOT))
from services.chat.session import ChatSession, IsolationError, load_world
WORLD = load_world(ROOT)
for who, tok, m in [("priya", "guess", "mandate_a"), ("mallory", "priya-fixture-token", "mandate_a"), ("priya", "priya-fixture-token", "mandate_z")]:
    try:
        ChatSession(who, tok, m, WORLD, ws=WS)
    except PermissionError as e:
        print(f"login {who}/{m}: refused ({e})")
chat = ChatSession("priya", "priya-fixture-token", "mandate_a", WORLD, ws=WS)
print("\nsession:", chat.session, "| a chat line is:", channel.authority(chat.message("hi"), USERS))
chat_ctx, _ = chat.context()
print("B canaries in A's chat context:", evals.canary_hits([chat_ctx], CAN["mandate_b"]), "| Northgate:", "Northgate" in chat_ctx)
reply, trace = chat.ask("What changed for my client after the release?")
print("\n" + reply)
print("\ntools called:", [(r["tool"], r.get("action_class")) for r in trace if r.get("tool")])
````
